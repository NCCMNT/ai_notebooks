## Imports and helper functions

```python
import numpy as np
from sklearn.metrics import root_mean_squared_error
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

```python
def get_synthetic_data(n: int = 100, noise: float = 0.1) -> tuple:
    X = np.random.rand(n,1)
    a_true = np.random.random()
    b_true = np.random.random()
    y = a_true * X + b_true + np.random.randn(n, 1) * noise

    print(f"True slope: {a_true:.4f}")
    print(f"True intercept: {b_true:.4f}")
    return X,y, a_true, b_true
```

```python
def split_dataset(X: np.ndarray, y: np.ndarray, split: float = 0.3) -> tuple:
    n = X.shape[0]
    n_test = int(n*split)

    indices = np.random.permutation(n)

    test_indices = indices[:n_test]
    train_indices = indices[n_test:]

    X_train, X_test = X[train_indices], X[test_indices]
    y_train, y_test = y[train_indices], y[test_indices]

    return X_train, X_test, y_train, y_test
```

# SimpleMLP Architecture (1-hidden-layer ReLU)

**Input:**  
$$
X \in \mathbb{R}^{n \times d}, \quad n - \text{samples}, \; d - \text{input dim}
$$

**Hidden layer (pre-activation and ReLU):**  
$$
Z_1 = X W_1 + b_1, \quad
A_1 = \text{ReLU}(Z_1) = \max(0, Z_1)
$$

**Output layer (regression):**  
$$
Z_2 = A_1 W_2 + b_2, \quad \hat{y} = Z_2
$$

# Introduction to Backpropagation

Backpropagation is the algorithm that allows neural networks to **learn**.  
It provides a systematic way to compute how each weight in the network should change in order to **reduce the loss**.

## What do we need to compute?

To train a neural network using gradient descent, every weight matrix $W$ and bias vector $b$ must be updated using the rule:

$$
W \leftarrow W - \eta \frac{\partial L}{\partial W}, 
\qquad
b \leftarrow b - \eta \frac{\partial L}{\partial b},
$$

where:

- $L$ is the loss function (e.g., mean squared error),
- $\eta$ is the learning rate,
- $\frac{\partial L}{\partial W}$ and $\frac{\partial L}{\partial b}$ are **gradients** telling us how much each parameter contributed to the final error.

So the main goal of backpropagation is to compute **all partial derivatives of the loss with respect to every parameter** in the network.

## How does backpropagation compute these gradients?

Backpropagation uses the **chain rule** from calculus to propagate the loss backward through the network.

A neural network computes output through layers:

$$
X \rightarrow \text{Layer 1} \rightarrow \text{Layer 2} \rightarrow \cdots \rightarrow \text{Output}.
$$

The loss is only computed at the output layer, but it depends on **all** earlier computations.  
Therefore we compute gradients from the output backwards:

1. Compute how the loss changes with respect to the output:
   $$
   \frac{\partial L}{\partial Z_{\text{output}}}
   $$

2. Use the chain rule to propagate this error back through each layer:
   $$
   \frac{\partial L}{\partial Z_{l}} 
   = 
   \frac{\partial L}{\partial Z_{l+1}}
   \cdot
   \frac{\partial Z_{l+1}}{\partial A_l}
   \cdot
   \frac{\partial A_l}{\partial Z_l}.
   $$

3. Once we know how the loss depends on each layer's activation, we can compute its parameter gradients:
   $$
   \frac{\partial L}{\partial W_l},
   \qquad
   \frac{\partial L}{\partial b_l}.
   $$

# Mean Squared Error (MSE) Loss

Given predictions $\hat{y}$ and true labels $y$:

$$
L = \frac{1}{m} \sum_{i=1}^m (\hat{y}_i - y_i)^2
$$

During backpropagation, we compute the partial derivatives with respect to all parameters.


# Backprop: Output Layer

MSE loss gradient:
$$
\delta_2 = \frac{\partial L}{\partial Z_2} = \frac{2}{m}(\hat{y}-y)
$$

Parameter gradients: 
$$
\frac{\partial L}{\partial W_2} = A_1^\top \delta_2, \quad
\frac{\partial L}{\partial b_2} = \sum_i \delta_2^{(i)}
$$

Gradient to hidden layer:
$$
\frac{\partial L}{\partial A_1} = \delta_2 W_2^\top
$$


# Backprop: Hidden Layer

Through ReLU:
$$
\frac{\partial L}{\partial Z_1} = \frac{\partial L}{\partial A_1} \odot \mathbf{1}_{Z_1>0}
$$

Parameter gradients:
$$
\frac{\partial L}{\partial W_1} = X^\top \frac{\partial L}{\partial Z_1}, \quad
\frac{\partial L}{\partial b_1} = \sum_i \frac{\partial L}{\partial Z_1^{(i)}}
$$

# Why we compute $\frac{\partial L}{\partial W_2}$ in backprop — and why it equals $(A_1^\top \cdot \frac{\partial L}{\partial Z_2}$)

Backpropagation is simply the application of **gradient descent** to neural networks.  
To update any weight matrix, we need its gradient:

$$
W_2 \leftarrow W_2 - \eta \frac{\partial L}{\partial W_2}.
$$

So we must compute:

$$
\frac{\partial L}{\partial W_2}.
$$

This tells us *how much each weight contributed to the loss* and how to adjust it.

## Using the chain rule

We want:

$$
\frac{\partial L}{\partial W_2}
=
\frac{\partial L}{\partial Z_2}
\cdot
\frac{\partial Z_2}{\partial W_2}.
$$

We already know:

$$
\frac{\partial L}{\partial Z_2}
= \frac{2}{m}(Z_2 - y),
$$

often called the output error:

$$
\delta_2 = \frac{\partial L}{\partial Z_2}.
$$

### Derivative of $Z_2$ with respect to $W_2$

Because:

$$
Z_2 = A_1 W_2 + b_2,
$$

consider one sample $i$:

$$
Z_2^{(i)} = A_1^{(i)} W_2.
$$

Derivative of $Z_2^{(i)}$ with respect to a weight $W_{2,j}$ is:

$$
\frac{\partial Z_2^{(i)}}{\partial W_{2,j}} = A_{1,j}^{(i)}.
$$

For all samples stacked together:

$$
\frac{\partial Z_2}{\partial W_2} = A_1.
$$

## Combine using matrix multiplication

Putting chain rule terms together:

$$
\frac{\partial L}{\partial W_2}
=
A_1^\top \delta_2.
$$

Shape check:

- $A_1^\top$: $h \times m$  
- $\delta_2$: $m \times 1$

Result:

$$
(h \times m)(m \times 1) = h \times 1,
$$

matching the shape of $W_2$.


# Backpropagation: Hidden Layer

Hidden pre-activation:
$$
Z_1 = XW_1 + b_1
$$

ReLU activation:
$$
A_1 = \max(0, Z_1)
$$

ReLU derivative:
$$
\frac{d}{dZ_1}\text{ReLU}(Z_1) =
\begin{cases}
1 & Z_1 > 0 \\
0 & Z_1 \le 0
\end{cases}
$$

### Gradient for hidden pre-activation:
$$
\frac{\partial L}{\partial Z_1}
=
\frac{\partial L}{\partial A_1}
\odot
\text{ReLU}'(Z_1)
$$

### Gradients for $W_1$ and $b_1$:
$$
\frac{\partial L}{\partial W_1}
    = X^\top \frac{\partial L}{\partial Z_1}
$$

$$
\frac{\partial L}{\partial b_1}
    = \sum_{i=1}^m \frac{\partial L}{\partial Z_1^{(i)}}
$$

```python
class SimpleMLP:
    def __init__(
            self,
            input_size: int = 1,
            hidden_size: int = 10,
            output_size: int = 1,
            lr: float = 0.01
        ):
        self.lr = lr

        self.W1 = np.random.randn(input_size, hidden_size) * 0.1
        self.b1 = np.zeros((1, hidden_size))
        
        self.W2 = np.random.randn(hidden_size, output_size) * 0.1
        self.b2 = np.zeros((1,output_size))
    
    def relu(self, x: np.ndarray) -> np.ndarray:
        return np.maximum(0,x)
    
    def relu_derivative(self, x: np.ndarray) -> np.ndarray:
        return (x > 0).astype(float) # 0 if x <= 0 and 1 if x > 0
    
    def forward(self, X: np.ndarray) -> np.ndarray:
        self.X = X                             # (n, input)
        self.z1 = X @ self.W1 + self.b1        # (n, hidden)
        self.a1 = self.relu(self.z1)           # (n, hidden)
        self.z2 = self.a1 @ self.W2 + self.b2  # (n, output)
        return self.z2
    
    def backward(self, y_true: np.ndarray, y_pred: np.ndarray) -> None:
        m = y_true.shape[0]

        # loss: MSE = sum((y - y_pred)**2) / m
        # y_pred = z2 = a1W2 + b2
        # dloss / dz2 = (2/m)(y_pred - y)
        dz2 = (y_pred - y_true) / m

        # dloss / dW2 = a1.T @ dloss / dz2
        # dloss / db2 = sum(dloss/dz2[i] for i)
        dW2 = self.a1.T @ dz2
        db2 = np.sum(dz2, axis=0, keepdims=True)

        # da1 = dz2 @ W2.T
        da1 = dz2 @ self.W2.T

        # dz1 = da1 * ReLU'(z1)
        dz1 = da1 * self.relu_derivative(self.z1)

        #z1 
        dW1 = self.X.T @ dz1
        db1 = np.sum(dz1, axis=0, keepdims=True)

        self.W2 -= self.lr * dW2
        self.b2 -= self.lr * db2
        self.W1 -= self.lr * dW1
        self.b1 -= self.lr * db1
    
    def fit(self, X: np.ndarray, y: np.ndarray, epochs: int = 5000, verbose: bool = True) -> None:
        for epoch in range(epochs):
            y_pred = self.forward(X)
            self.backward(y, y_pred)

            if verbose and epoch % (epochs // 10) == 0:
                loss = np.mean((y_pred - y) ** 2)
                print(f"Epoch {epoch}, Loss (MSE): {loss:.4f}")
    
    def predict(self, X: np.ndarray) -> np.ndarray:
        return self.forward(X)
```

```python
X,y, a_true, b_true = get_synthetic_data(1000)
X_train, X_test, y_train, y_test = split_dataset(X,y)
```

```python
fig, ax = plt.subplots(figsize=(12,8))
ax.scatter(X_train, y_train, color='b', alpha=0.5, label='train')
ax.scatter(X_test, y_test, color='c', alpha=0.5, label='test')
ax.set_xlabel('X')
ax.set_ylabel('y')
ax.set_title(f'Regression line evolution ({a_true:.4f}x + {b_true:.4f})')

line, = ax.plot([], [], 'r-', lw=2)  # empty line object to update
model = SimpleMLP(lr=0.02)
epochs_per_frame = 100
plt.close()
```

```python
def init():
    line.set_data([], [])
    return line,

def update(frame_idx):
    model.fit(X_train, y_train, epochs=epochs_per_frame, verbose=False)
    
    # Regression line
    X_line = np.linspace(X_train.min(), X_train.max(), 100).reshape(-1,1)
    y_line = model.predict(X_line)
    
    line.set_data(X_line.flatten(), y_line.flatten())
    
    # RMSE for legend
    y_pred_train = model.predict(X_train)
    y_pred_test = model.predict(X_test)
    
    rmse_train = root_mean_squared_error(y_train, y_pred_train)
    rmse_test = root_mean_squared_error(y_test, y_pred_test)
    
    lin_fit = LinearRegression().fit(X_line, y_line)
    slope = lin_fit.coef_[0][0]       # approximate slope
    intercept = lin_fit.intercept_[0] # approximate intercept
    total_epochs = (frame_idx+1) * epochs_per_frame

    label = f'Epoch {total_epochs} (RMSE train: {rmse_train:.4f}, test: {rmse_test:.4f})\nApproximate slope: {slope:.4f}, intercept: {intercept:.4f}'

    line.set_label(label)
    ax.legend()
    
    return line,

ani = FuncAnimation(fig, update, frames=60, init_func=init, blit=True, interval=200)
plt.close()

HTML(ani.to_jshtml())
```
